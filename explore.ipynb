## 1. Clean your Datasets

For the first part of your project, you will apply the same data cleaning steps to your "data/control.txt" and "data/treatment.txt" files. By the end of this step, you should have two lists of floats that match the numbers you have in each respective file.

You will implement your `clean_data` function in your `cleaner.py` module and import the module in the code-block below.

```python
# TODO: import all the packages that you will need for this notebook here
import cleaner
import statistics as stat
```

```python
# TODO: read your text files into your code using file I/O
control_data =[]
treatment_data = []
cleaned_control_data =[]
cleaned_treatment_data = []

with open("data/control.txt") as control:
    control_data = control.readlines()

with open("data/treatment.txt") as treatment:
    treatment_data = treatment.readlines()

# Test print statememts 
#print(treatment_data)
#print(control_data)

# TODO: clean both list objects that  you created above using the 'clean_data' function
cleaned_control_data = cleaner.clean_data(control_data)
cleaned_treatment_data = cleaner.clean_data(treatment_data)

print ("Control Data: ",cleaned_control_data)
print ("Treatment Data:", cleaned_treatment_data)
```

## 2. Explore your Datasets

Now that you have your list objects created and cleaned, let's explore the descriptive statistics of these two files and make subsequent observations on our data.

```python
# TODO: Calculate the standard deviation of both the control and treatment group
std_control_data = stat.stdev(cleaned_control_data)
std_treatment_data = stat.stdev(cleaned_treatment_data)
# TODO: print your findings
print("Standard Deviation of Control Data:", std_control_data)
print("Standard Deviation of Treatment Data",  std_treatment_data)
```

```python
# TODO: Calculate the mean of both the control and treatment group
mean_control_data = stat.mean(cleaned_control_data)
mean_treatment_data = stat.mean(cleaned_treatment_data)

# TODO: print your findings
print("Mean of Control Data" ,mean_control_data)
print("Mean of Treatment Data" , mean_treatment_data)
```

```python
# TODO: Are there any outliers in your control group? (use 2 * stdev to find these outliers)
outliers_control = []
# lower_bound_control= mean_control_data - (2* std_control_data)
# upper_bound_control= mean_control_data + (2* std_control_data)
# outliers_control = [x for x in cleaned_control_data]
#print(lower_bound_control ,upper_bound_control)
for x in cleaned_control_data:
    if x < mean_control_data-2* std_control_data or x > mean_control_data + 2* std_control_data: 
        outliers_control.append(x)

print("Outliers Control Data:", outliers_control)
print(len(outliers_control))

```

```python
# TODO: Are there any outliers in your treatment group?
outliers_treatment = []
# lower_bound= mean_treatment_data - (2* std_treatment_data)
# upper_bound= mean_treatment_data + (2* std_treatment_data)
# outliers_control = [x for x in cleaned_treatment_data]
# print(lower_bound, upper_bound)

for x in cleaned_treatment_data:
    if x < mean_treatment_data-2* std_treatment_data or x > mean_treatment_data + 2* std_treatment_data: 
        outliers_treatment.append(x)

print("Outliers Treatment Data:", outliers_treatment)
print(len(outliers_treatment))
```
